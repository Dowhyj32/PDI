## **¿Que es una Imagen?**

En el mundo de **Procesamiento de Imágenes**, una **imagen** puede ser vista como un **array de píxeles** en dos dimensiones. Un Píxel es la unidad más pequeña de una imagen, y su valor dependera de la representación de ella.

1.  **Imágenes en Blanco y Negro (Escala de Grises):** Cada píxel tiene un solo valor que representa su intensidad luminosa. Este valor suele ser un número entre 0 (negro) y 255 (blanco).
    
2.  **Imágenes a Color (RGB):**  Cada píxel está compuesto por tres valores, uno por cada canal de color: **Rojo**, **Verde** y **Azul**. Cada valor de estos canales puede variar entre 0 y 255, lo que permite representar una amplia gama de colores.
    

Para ilustrar este concepto, imagina que tienes una imagen **M**:

-   En una imagen en blanco y negro, `M[a][b]` te diría el valor de luminancia del píxel en la fila `a` y la columna `b`.
-   En una imagen RGB, `M[a][b][0]` representa la intensidad del canal **Rojo** en el píxel ubicado en la fila `a` y columna `b`. Igualmente, `M[a][b][1]` sería la intensidad del canal **Verde** y `M[a][b][2]` la del **Azul**.

### ¿Y cómo se implementa esto en **Scikit-Image**?

**Scikit-Image** utiliza arrays de **Numpy** para representar las imágenes. Esta estructura tiene varias ventajas:

-   **Indexación eficiente**: Gracias a Numpy, podemos acceder a los píxeles de manera rápida y eficaz.
-   **Operaciones vectorizadas**: Podemos aplicar transformaciones o filtros a toda la imagen sin tener que escribir ciclos complicados.
-   **Funciones especializadas**: Numpy ofrece muchas funciones que hacen que el procesamiento de imágenes sea más sencillo, como operaciones de álgebra lineal, transformación de matrices, mascaras, etc.

```python
# Cargamos una imagen en blanco y negro
imagenBlancoYNegro = data.camera()

# Tanto de la forma individual como por rango, podemos setear su valor
# Creamos una copia y seteamos en negro a los pixeles en filas pares y columnas impares
imagenCopia = imagenBlancoYNegro.copy()
imagenCopia[::2, 1::2] = 1

fig, (ax1, ax2) = plt.subplots(1,2)
ax1.set_title("Camera Man Original")
ax1.axis("off")
ax1.imshow(imagenBlancoYNegro, cmap=plt.cm.gray)

ax2.set_title("Camera Man Modificada")
ax2.axis("off")
ax2.imshow(imagenCopia, cmap=plt.cm.gray)
plt.show()
```

### Mascaras




Al hablar de mascaras no nos referimos a los 'Masked Arrays' de Numpy que permiten crear arreglos con informacion faltante o invalida, sino a la capacidad de crear arreglos booleanos segun una condicion sobre los valores de la imagen.

```python
# Cargamos una imagen en blanco y negro
imagenBlancoYNegro = data.camera()

# Podemos evaluar una condicion sobre los valores de la imagen
# Consiguiendo un Arreglo Booleano que tendra True en donde sea verdadera
# y False donde no.

# Conseguimos un arreglo booleano sobre los valores que sean mayor al promedio
# de la imagen
mascara = imagenBlancoYNegro > np.mean(imagenBlancoYNegro)

# Podemos utilizar una mascara para Indexar a los pixeles que cumplan la condicion

# Creamos una copia de la imagen y oscurecemos las zonas brillantes de la imagen
imagenOscurecida = imagenBlancoYNegro.copy()
imagenOscurecida[mascara] -= 100

print(f"Es tipo de dato de la mascara es: {mascara.dtype}")

fig, (ax1, ax2, ax3) = plt.subplots(1,3, figsize=(20,20))
ax1.set_title("Camera Man Original")
ax1.axis("off")
ax1.imshow(imagenBlancoYNegro, cmap=plt.cm.gray)

ax2.set_title("Mascara")
ax2.axis("off")
ax2.imshow(mascara, cmap=plt.cm.gray)

ax3.set_title("Imagen Oscurecida")
ax3.axis("off")
ax3.imshow(imagenOscurecida, cmap=plt.cm.gray)
plt.show()
```

```python
# Con las mascaras tambine podemos crear una zona de interes

# Cargamos una imagen en blanco y negro, guardamos su alto y ancho
imagenBlancoYNegro = data.camera()
nrows, ncols = imagenBlancoYNegro.shape

# Creamos un arreglo de (alto, 1) y (1, ancho)
# donde cada valor es el correspondiente index
rows, cols = np.ogrid[:nrows, :ncols]

# De esta forma podemos crear condiciones sobre las posiciones de los pixeles

# Creamos una mascara para todos los pixeles que no esten dentro de un circulo de radio 100
centroRow = nrows / 2
centroCol = ncols / 2
afueraDiscoMascara = (rows - centroRow)**2 + (cols - centroCol)**2 > 100**2

# Seteamos en 0 a todos estos pixeles
imagenBlancoYNegro[afueraDiscoMascara] = 0

fig, (ax1, ax2) = plt.subplots(1,2)
ax1.set_title("Mascara")
ax1.axis("off")
ax1.imshow(afueraDiscoMascara, cmap=plt.cm.gray)

ax2.set_title("Camera Man Modificada")
ax2.axis("off")
ax2.imshow(imagenBlancoYNegro, cmap=plt.cm.gray)
plt.show()
```

## **Tipos de datos en Scikit-Image**

Cuando trabajamos con imágenes, la mayoría de las veces estamos acostumbrados a representar los valores de los píxeles utilizando enteros entre **0** y **255** (tipo de dato `uint8`). Sin embargo, **Scikit-Image** (y en general **Numpy**) es mucho más flexible y soporta una variedad de tipos de datos (**dtypes**) para representar imágenes. Esto es crucial porque, dependiendo de la operación que estemos haciendo, el rango de valores de los píxeles puede variar.

El rango de valores valido que puede manejar cada representacion es:

| **Tipo de dato** | **Rango de valores**                    |
|---|---|
| `uint8`          | 0 a 255                                 |
| `uint16`         | 0 a 65535                               |
| `uint32`         | 0 a 2<sup>32</sup> - 1                 |
| `float`          | -1 a 1 o 0 a 1 |
| `int8`           | -128 a 127                              |
| `int16`          | -32768 a 32767                          |
| `int32`          | -2<sup>31</sup> a 2<sup>31</sup> - 1   |

### ¿Por qué es importante esto?

-   Los tipos de datos permiten que **Scikit-Image** maneje imágenes con diferentes rangos de valores según el contexto. Por ejemplo, las imágenes en **escala de grises** pueden usar `uint8` para representar los valores de intensidad, pero si estamos trabajando con imágenes científicas o de mayor precisión, tal vez queramos usar `uint16` o incluso `float` para representar mejor los valores de los píxeles.
    
-   Las funciones en **Scikit-Image** aceptaran imágenes en cualquiera de estos tipos de datos, pero es importante tener en cuenta que **las funciones no siempre devuelven imágenes con el mismo tipo de dato que la entrada**. Esto significa que, por ejemplo, si tomamos una imagen en formato `uint8` y aplicamos un filtro, el resultado podría ser una imagen en formato `float`, especialmente si la operación involucra normalización o cálculos que requieren mayor precisión.

```python
from skimage import data, filters,color, feature
import numpy as np
import matplotlib.pyplot as plt

# Cargamos una imagen Blanco y Negro de ejemplo en formato uint8
imagen = data.cell()
#imagen = color.rgb2gray(imagen)
#imBN = color.rgb2gray(imagen)

# Verificamos el tipo de Dato
print(f"Tipo de dato original: {imagen.dtype}")
print(f"Minimo: {np.min(imagen)}")
print(f"Maximo: {np.max(imagen)}")

# Aplicamos un filtro de Skimage
imagenFiltrada = filters.sobel(imagen)

print()
print(f"Tipo de dato después de aplicar filtro: {imagenFiltrada.dtype}")
print(f"Minimo: {np.min(imagenFiltrada)}")
print(f"Maximo: {np.max(imagenFiltrada)}")

# Mostrar las imágenes
# (Vemos que cuando los rangos son correctos, se muestran correctamente)
fig, ax = plt.subplots(1, 2, figsize=(10, 5))
ax[0].set_title('Imagen uint8')
ax[0].axis('off')
ax[0].imshow(imagen, cmap='gray')

ax[1].set_title('Imagen con Filtro')
ax[1].axis('off')
ax[1].imshow(imagenFiltrada, cmap='gray')

plt.show()
```

### Conversiones de Tipos de Datos

En caso de querer un dtype particular, skimage ofrece [funciones](https://scikit-image.org/docs/0.23.x/api/skimage.util.html) para realizar conversiones de tipo, reescalando correctamente los valores de intensidad. Nunca utilicen `astype` en una imagen, porque podrian generar valores fuera del rango valido durante la conversion.

```python
from skimage import data, filters, util
import numpy as np
import matplotlib.pyplot as plt

# Cargamos una imagen Blanco y Negro de ejemplo en formato uint8
imagen = data.camera()

# Verificamos el tipo de Dato
print(f"Tipo de dato original: {imagen.dtype}")
print(f"Minimo: {np.min(imagen)}")
print(f"Maximo: {np.max(imagen)}")

# Cambiamos el tipo de dato a float, reescalando los valores correctamente
imagenFloat = util.img_as_float(imagen)

print()
print(f"Tipo de dato después conversion: {imagenFloat.dtype}")
print(f"Minimo: {np.min(imagenFloat)}")
print(f"Maximo: {np.max(imagenFloat)}")

# Mostrar las imágenes
# (Vemos que cuando los rangos son correctos, se muestran correctamente)
fig, ax = plt.subplots(1, 2, figsize=(10, 5))
ax[0].set_title('Imagen uint8')
ax[0].axis('off')
ax[0].imshow(imagen, cmap='gray')

ax[1].set_title('Imagen float')
ax[1].axis('off')
ax[1].imshow(imagenFloat, cmap='gray')

plt.show()
```

```python
from skimage import data, filters, util, feature
import numpy as np
import matplotlib.pyplot as plt

def printInfo(imagen):
  print(f"Tipo de dato original: {imagen.dtype}")
  print(f"Minimo: {np.min(imagen)}")
  print(f"Maximo: {np.max(imagen)}")
  print()

# Cargamos una imagen Blanco y Negro de ejemplo en formato uint8
imagen = data.camera()
printInfo(imagen)

# Realizamos la conversion correcta a tipo float
# Notar que hubo un reescalado correcto de los datos
imagenFloatValida = util.img_as_float(imagen)
printInfo(imagenFloatValida)

# Realizamos la conversion incorrecta al tipo float
# Notar notar que no hubo ningun reescalado de los datos
imagenFloatInvalida = imagen.astype(float)
printInfo(imagenFloatInvalida)

# Aplicamos el algoritmo de Canny
# utilizado para encontrar los bordes de una imagen
# tanto en la imagen con valores validos como en la de invalidos
imagenValBordes = feature.canny(imagenFloatValida, sigma=2)
imagenInvalBordes = feature.canny(imagenFloatInvalida, sigma=2)

fig, ax = plt.subplots(1, 3, figsize=(20, 20))
ax[0].set_title('Imagen Original')
ax[0].axis('off')
ax[0].imshow(imagen, cmap='gray')

ax[1].set_title('Bordes Imagen Valida')
ax[1].axis('off')
ax[1].imshow(imagenValBordes, cmap='gray')

ax[2].set_title('Bordes Imagen Invalida')
ax[2].axis('off')
ax[2].imshow(imagenInvalBordes, cmap='gray')

plt.show()
```

Por convesion, al pasar de un tipo de dato con signo a uno sin signo, todos los valores negativos iran automaticamente a 0.

```python
from skimage import data, util
import numpy as np

# Creamos un arreglo de tipo int8 con valores positivos y negativos.
arregloConNegativos = np.asarray([[-128, -75, -20],[0, 50, 75],[100, 115, 127]], dtype=np.int8)
print(f"Arreglo Original en int8: \n {arregloConNegativos}")

# Se reescalan los valores positivos y los negativos quedan en 0
arregloUint8 = util.img_as_ubyte(arregloConNegativos)

print()
print(f"Arreglo Original en Uint8: \n {arregloUint8}")
```

### Limitaciones de los tipos de datos

Es importante asegurarse que las operaciones que hagamos sobre una imagen sean válidas sobre el tipo que se está utilizando, de otra manera podríamos encontrarnos con comportamientos inesperados. Tomen por ejemplo el caso de un dato con valor 0 del tipo `uint8`. Sabemos que su rango va de 0 a 255. ¿Qué pasaría si restáramos este dato por 1? El resultado es un *overflow* del rango representable, haciendo que el valor luego de la resta sea 255.

```python
from skimage import data, filters, util, feature
import numpy as np
import matplotlib.pyplot as plt

# Creamos un valor de tipo uint8 con valor 0
valorUint8 = np.asarray([0], dtype=np.uint8)

print(f"Tipo de la variable: {valorUint8.dtype}")
print(f"Valor de la variable: {valorUint8[0]}")

# Restamos el valor por 1, -1 no se encuentra en el rango posible
# Sucede un overflow y termina en el valor maximo 255
valorUint8[0] -= 1

print(f"Valor de la variable luego de la resta: {valorUint8[0]}")
```

```python
# Esto tambien puede suceder en una imagen

from skimage import data
import numpy as np
import matplotlib.pyplot as plt

# Cargamos una imagen en grises de https://scikit-image.org/docs/0.25.x/api/skimage.data.html formato uint8
imagen = data.camera()

# Creamos una copia y restamos todos sus valores por 100
imagenRestada = imagen.copy() - 100

# Veamos que las zonas mas oscuras de la imagen original ahora son blancas
# Causa del overflow en el tipo de dato.

fig, ax = plt.subplots(1, 2, figsize=(20, 20))
ax[0].set_title('Imagen Original')
ax[0].axis('off')
ax[0].imshow(imagen, cmap='gray')

ax[1].set_title('Imagen Restada')
ax[1].axis('off')
ax[1].imshow(imagenRestada, cmap='gray')

plt.show()

%%%TAREA: repita el experimento para
```

## **Extender o Achicar el Rango de Valores**

Aunque no sea recomendado, hay veces en donde tendremos que modificar nuestra imagen para extenderse completamente sobre un rango que nosotros necesitemos. Para esto utilizaremos la función `skimage.exposure.rescale_intensity` que nos permitirá definir el rango de entrada y de salida.

[Documentacion](https://scikit-image.org/docs/0.23.x/api/skimage.exposure.html#skimage.exposure.rescale_intensity)

```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import exposure
from skimage import data

# Cargamos una imagen de ejemplo
imagen = data.camera()

fig, axs = plt.subplots(2,2, figsize=(10,10))
axs = axs.ravel()
axs[0].axis("off")
axs[0].set_title("Imagen Original")
axs[0].imshow(imagen, cmap='gray')

axs[1].set_title("Histograma Original")
axs[1].set_xlim(0, 255)
axs[1].hist(imagen.ravel(), bins=256, histtype='step', color='black')

# Usamos rescale_intensity para ajustar la imagen al rango [0, 100]
imagenRescaled = exposure.rescale_intensity(imagen, in_range=(imagen.min(), imagen.max()), out_range=(0, 100))

axs[2].axis("off")
axs[2].set_title("Imagen Reescalada")
axs[2].imshow(imagenRescaled, cmap='gray', vmin=0, vmax=255)

axs[3].set_title("Histograma Reescalado")
axs[3].set_xlim(0, 255)
axs[3].hist(imagenRescaled.ravel(), bins=256, histtype='step', color='black')

plt.tight_layout()
plt.show()
```

## **Cómo Visualizar Imágenes con Matplotlib**

https://matplotlib-org.translate.goog/stable/users/explain/quick_start.html?_x_tr_sl=en&_x_tr_tl=es&_x_tr_hl=es&_x_tr_pto=tc

Para visualizar imágenes en Python, una de las bibliotecas más utilizadas es **Matplotlib**. Además de ser una herramienta poderosa y flexible, se integra perfectamente con **Scikit-Image**, lo que facilita el proceso de visualización y manipulación de imágenes.

Antes de empezar con los ejemplos de funciones y gráficos, es importante entender los tres objetos clave en **Matplotlib** que forman la jerarquía básica de la visualización:

### 1. **Figure**
La **Figure** es el contenedor principal donde se muestra todo el gráfico o la imagen final. Es la ventana o el área donde se dibujan uno o más **Axes**. Piensen en ella como el canvas completo donde todo se organiza y visualiza.

### 2. **Axes**
Cada **Axes** es una región dentro de la **Figure** que contiene la visualización propiamente dicha. Dentro de los **Axes**, se dibujan los elementos gráficos (como imágenes, gráficos de líneas, barras, etc.). Cada **Axes** puede tener de 2 a 3 **Axis**, dependiendo de si se trata de un gráfico bidimensional o tridimensional.

Los **Axes** son donde se configura la mayor parte de las propiedades visuales, como el tamaño, la escala, los títulos, las leyendas, y los ejes.

### 3. **Axis**
Los **Axis** son los ejes dentro de cada **Axes**, y son responsables de definir la escala, los límites, y la ubicación de las marcas o *ticks*. Estos ejes permiten controlar el aspecto de las escalas de los datos, y cómo se muestran los valores o las etiquetas en los ejes.

### Resumen de la jerarquía:

1. **Figure**: Contenedor general que puede tener uno o más **Axes**.
2. **Axes**: Área dentro de la **Figure** donde se visualiza la información.
3. **Axis**: Los ejes dentro de cada **Axes** que controlan la escala y los valores.

### Visualización de la jerarquía

Para entender mejor cómo se estructuran estos objetos en **Matplotlib**, puedes consultar la siguiente imagen que muestra cómo se distribuyen estos elementos dentro de una visualización:

![Anatomía de Matplotlib](https://matplotlib.org/stable/_images/anatomy.png)
---

Con estos conceptos básicos, ya podemos comenzar a crear gráficos, modificar sus propiedades, y explorar más sobre los atributos personalizables que **Matplotlib** ofrece.


```python
import matplotlib.pyplot as plt
import numpy as np

# Creamos una figura con una grilla de 2x2 de ejes
fig, axs = plt.subplots(2, 2, figsize=(10, 8))  # Especificamos el tamaño de la figura en pulgadas

"""
Viendo el último ejemplo, el orden de los ejes en la visualización será el siguiente:

  axs[0,0]    axs[0,1]

  axs[1,0]    axs[1,1]

Podemos usar `ravel()` para aplanar el arreglo y hacer más fácil referenciarlos
"""

# Aplanamos el arreglo para facilitar el acceso a los subgráficos
axs = axs.ravel()

# Definimos un título para la figura completa
fig.suptitle('Visualización de Varios Gráficos', fontsize=18, fontweight='bold')

# Primer gráfico: Línea simple
axs[0].set_title("Gráfico de Línea", fontsize=14)
axs[0].axis("off")  # Escondemos los ejes
axs[0].plot([1, 2, 3, 4], [1, 4, 2, 3], marker='o', color='b', linestyle='-', linewidth=2, markersize=8)

# Segundo gráfico: Histograma
axs[1].set_title("Histograma", fontsize=14)
axs[1].set_xlabel("Valores", fontsize=12)
axs[1].set_ylabel("Frecuencia", fontsize=12)
axs[1].hist(np.random.randn(1000), bins=30, color='g', alpha=0.7)
axs[1].grid(True)

# Tercer gráfico: Dispersión con leyendas
axs[2].set_title("Gráfico de Dispersión", fontsize=14)
axs[2].scatter([1, 23, 2, 4, 5], [12, 23, 3, 10, 1], label='Grupo 1', color='r')
axs[2].scatter([5, 14, 66, 1, 8], [52, 2, 56, 20, 6], label='Grupo 2', color='b')
axs[2].set_xlabel("Eje X", fontsize=12)
axs[2].set_ylabel("Eje Y", fontsize=12)
axs[2].legend(loc='upper right')
axs[2].grid(True)

# Cuarto gráfico: Gráfico de línea con múltiples conjuntos de datos
axs[3].set_title("Gráfico de Línea", fontsize=14)
axs[3].plot([1, 2, 3, 4], [1, 2, 3, 4], label="Conjunto 1", color='orange', linestyle='--', linewidth=2)
axs[3].plot([1, 2, 3, 4], [4, 3, 2, 1], label="Conjunto 2", color='purple', linestyle='-.', linewidth=2)
axs[3].set_xlabel("Eje X", fontsize=12)
axs[3].set_ylabel("Eje Y", fontsize=12)
axs[3].legend(loc='upper left')
axs[3].grid(True)

# Ajustamos el espaciado entre los subgráficos
plt.tight_layout(rect=[0, 0, 1, 0.96])  # Para evitar que se solapen los títulos
plt.show()
```

### **Plotear Imagenes**
Para poder visualizar imagenes usaremos la funcion `imshow `, la cual nos permitira mostrar imagenes en un axes, pudiendo definir un mapa de colores y su rango de interes.

[Documentacion](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.imshow.html)

```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import exposure
from skimage import data

# Cargamos una imagen de ejemplo
imagen = data.camera()

fig, axs = plt.subplots(3,2, figsize=(10,10))
axs = axs.ravel()

# Mostramos una imagen y elegimos su mapa de colores a 'gray'
# Por default, el mapa de colores se mapeara a lo largo del rango de la imagen
# Es decir, min(imagen) se mostrara como negro y max(imagen) como blanco
# Aunque estos valores no sean 0 o 255.
axs[0].imshow(imagen, cmap='gray')
axs[0].set_title("Imagen Original")
axs[0].axis("off")

#mostramos el rango de valores
axs[1].set_title("Histograma Original")
axs[1].set_xlim(0, 255)
axs[1].hist(imagen.ravel(), bins=256, histtype='step', color='black')

# Usamos rescale_intensity para ajustar la imagen al rango [0, 100]
imagenRescaled = exposure.rescale_intensity(imagen, in_range=(imagen.min(), imagen.max()), out_range=(0, 100))

# Aunque el rango de valores cambio, no lo podemos ver visualmente
# No especificamos una zona de interes.
axs[2].imshow(imagenRescaled, cmap='gray')
axs[2].set_title("Imagen Reescalada")
axs[2].axis("off")

axs[3].set_title("Histograma Reescalado")
axs[3].set_xlim(0, 255)
axs[3].hist(imagenRescaled.ravel(), bins=256, histtype='step', color='black')

# Mostramos otra vez la imagen, pero esta vez definiendo explicitamente el rango del mapa de colores
# Definiendo 0 como el valor mas chico y 255 como mas grande
axs[4].imshow(imagenRescaled, cmap='gray', clim=(0,255))
axs[4].set_title("Imagen Reescalada con Rango Explicito")
axs[4].axis("off")

axs[5].set_title("Histograma Reescalado")
axs[5].set_xlim(0, 255)
axs[5].hist(imagenRescaled.ravel(), bins=256, histtype='step', color='black')


plt.tight_layout()
plt.show()
```

### Color Maps en Visualización de Imágenes

Un **Color Map** (o mapa de colores) es una función que asigna un rango de valores numéricos a un espacio de colores. En **Matplotlib**, existen diferentes tipos de mapas de colores, y cada uno está diseñado para representar de manera efectiva distintos tipos de información visual. Los mapas de colores se agrupan en 4 categorías según el tipo de datos que representan:

#### 1. **Sequential (Secuenciales)**:
Estos mapas de colores cambian la luminancia y saturación en un solo tono. Son útiles para representar información con una relación de orden, como la intensidad de una señal o la cantidad de un fenómeno. Un ejemplo común de este tipo es el mapa de colores **'grey'**, que va desde blanco (valor bajo) hasta negro (valor alto).

- **Ejemplo**: Uso en imágenes de intensidad o mapas de calor donde el valor de cada píxel sigue un orden lógico (por ejemplo, temperaturas, densidad, etc.).

#### 2. **Diverging (Divididos)**:
Los mapas de colores divergentes tienen dos tonos que se encuentran en el medio (a menudo blanco o un color neutro), y los extremos representan valores máximos o mínimos. Son ideales para mostrar datos que tienen una referencia central (como 0) y valores que se alejan de esta referencia. Un ejemplo es el mapa **'seismic'**.

- **Ejemplo**: Representar imágenes donde los valores negativos y positivos tienen diferentes significados, como en la visualización de errores o desviaciones respecto a una media.

#### 3. **Cyclic (Cíclicos)**:
Este tipo de mapas de colores usa dos o más tonos distintos en los extremos del espectro y se repiten de manera cíclica. Son útiles cuando los datos tienen una naturaleza cíclica, como los ángulos o tiempos que se repiten en un ciclo (por ejemplo, las horas del día). El mapa **'hsv'** es un buen ejemplo.

- **Ejemplo**: Usado para representar información en un rango circular, como ángulos, direcciones o cualquier valor periódico.

#### 4. **Qualitative (Cualitativos)**:
Los mapas de colores cualitativos usan colores que no tienen una relación de orden entre sí. Son útiles cuando se quiere representar diferentes categorías o clases sin que exista una relación cuantitativa entre ellas. Un ejemplo es el mapa de colores **'Pastel1'**.

- **Ejemplo**: Ideal para clasificar o diferenciar objetos o categorías dentro de una imagen sin que haya un orden o jerarquía entre ellas.

#### Elección del Color Map:
Dependiendo de la imagen que desees representar, es importante elegir el mapa de colores que mejor se adapte a la naturaleza de los datos y la historia que deseas contar con la visualización.

[Para conocer el resto de los Color Maps disponibles](https://matplotlib.org/stable/users/explain/colors/colormaps.html)

```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import data
from skimage import exposure

# Cargamos una imagen de ejemplo
image = data.camera()

# Conseguimos la diferencia de cada pixel sobre el valor promedio
mean_value = np.mean(image)
diff_image = image - mean_value

# Visualizamos la imagen de diferencias utilizando un mapa de colores divergente (seismic)
fig, ax = plt.subplots()
ax.set_title('Mapa Divergente: Diferencia respecto a la Media')
ax.axis('off')
ploteo = ax.imshow(diff_image, cmap='seismic', vmin=-np.max(np.abs(diff_image)), vmax=np.max(np.abs(diff_image)))
# Definimos una barra de colores para saber a que se relaciona cada color
plt.colorbar(ploteo)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import data, filters, color

# Cargamos una imagen de ejemplos
image = data.coins()

# Aplicamos un filtro de gradiente para obtener una imagen direccional
gradient_x = filters.sobel_h(image)  # Gradiente en la dirección horizontal
gradient_y = filters.sobel_v(image)  # Gradiente en la dirección vertical

# Calculamos la dirección de los gradientes (en radianes)
gradient_direction = np.arctan2(gradient_y, gradient_x)

# Visualisamos la imagen de dirección de gradiente utilizando un mapa de colores cíclico ('hsv')
fig, ax = plt.subplots()
ploteo=ax.imshow(gradient_direction, cmap='hsv')
ax.set_title('Mapa Cíclico: Direcciones de Gradiente')
ax.axis('off')
plt.colorbar(ploteo)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import data
from skimage import segmentation

# Cargamos una imagen de ejemplo
image = data.astronaut()

# Aplicamos un algoritmo de segmentación para crear categorías
segments = segmentation.slic(image, n_segments=40, compactness=4)

# Visualisamos las diferentes categorías utilizando un mapa de colores cualitativo (Pastel1)
fig, ax = plt.subplots()
ploteo = ax.imshow(segments, cmap='Pastel1')
ax.set_title('Mapa Cualitativo: Segmentación de Imágenes')
ax.axis('off')
plt.colorbar(ploteo)
plt.show()
```
